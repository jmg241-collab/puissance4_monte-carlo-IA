```python

import numpy as np
import time # Importe le module time pour la chronométrie
NOMBRE_DE_LIGNE = 6
NOMBRE_DE_COLONNE = 7
ROUGE = 1
JAUNE = 2
NOMBRE_DE_PARTIE_ALEATOIRE = 125


def generer() :
  '''
  none --> liste(array(entier))
  renvoie un plateau de jeu vide (que des O partout)
  '''
  liste_colonne = []
  liste_de_zero = []

  for i in range(NOMBRE_DE_COLONNE) :
    liste_de_zero = np.zeros(6, dtype=int)
    liste_colonne.append(liste_de_zero)

  liste_colonne = np.array(liste_colonne)

  return liste_colonne


#test de generer
Un_plateau = generer()
#print(Un_plateau)
#fin test



def genererbis():
    '''
    none --> array(array(entier))
    Renvoie un plateau de jeu vide (que des O partout).(plus simple que generer)
    '''
    return np.zeros((NOMBRE_DE_COLONNE, NOMBRE_DE_LIGNE), dtype=int)

def afficher_plateau(Un_plateau):
    M = Un_plateau.T
    M = np.flipud(M)
    LT = [" | ".join(map(str, vecteur_ligne))for vecteur_ligne in M]
    LL = len(f"| {LT[0]} |")
    S = "-" * LL
    LF = [f"| {ligne} |" for ligne in LT]
    print(f"\n{S}\n".join(LF))
    print(S)

#un_plateau = generer()

#un_plateau[3][0] = ROUGE  # Colonne 3, tout en bas
#un_plateau[3][1] = JAUNE  # Colonne 3, par dessus le rouge
#un_plateau[0][0] = JAUNE  # Colonne 0, tout en bas

# 3. Affichage
#print("Plateau:")
#print(un_plateau)
#print("PLateau Puissance 4 :")
#afficher_plateau(un_plateau)

#def nombre_de_joueur(plateau,k):
    #nombre=0
    #for i in plateau:s
        #nombre=nombre+i.count(k)
    #return nombre
    #return np.count_nonzero(plateau == k)

def colonne_pleine(colonne) :
  '''
  array(entier) --> booleen
  vérifie si une colonne est pleine, c'est-à-dire si elle contient encore une case vide.
  '''
  if 0 in colonne :
    return False
  return True


def joueur_suivant(plateau) :
  '''
  array(array(entier)) --> entier
  prend un plateau de jeu en entrée et renvoie la couleur du prochain joueur.
  en partant du principe qu'on commence par les rouges
  '''
  if nombre_de_joueur(plateau,1)==nombre_de_joueur(plateau,2) :
    return ROUGE
  return JAUNE

def plateau_rempli(plateau) :
  for colonne in plateau :
    if not(colonne_pleine(colonne)) :
      return False
  return True

def mouvement_possible(k,plateau) :
  '''entier,array(array(entier)) --> list(plateau)
  prend un plateau de jeu et un joueur (une couleur) en entrée et renvoie
  tous les mouvements possibles (c'est-à-dire une liste de plateaux).'''
  tableau_de_position = []
  for i in range(len(plateau)) :

    compteur = 0 # le compteur sert à vérifier si on a déjà traité une colonne ou pas. il vaut 1 si oui, 0 sinon.
    if not(colonne_pleine(plateau[i])) :

      for j in range(len(plateau[i])) :

        if plateau[i][j] == 0 :
          plateau_aux = np.copy(plateau) # crée une copie pour éviter de modifier le plateau d'origine
          plateau_aux[i][j] = k
          tableau_de_position.append(plateau_aux)
          compteur += 1  # on incrémente le compteur car on est entré dans le if donc on traite une colonne i.e. on ajoute un point pour créer une nouvelle position
        if j != 0 and compteur == 1 : # dès que l'on modifie une case différente de la première, on change de colonne
          break
        elif compteur == 1 : # si le compteur vaut 1, on a déjà traité la colonne donc on sort de la boucle for pour passer à la colonne suivante
          break
  return tableau_de_position

def gagnant(plateau):
    '''
    array(array(entier)) --> entier
    Parcourt le plateau pour voir si 4 pions sont alignés.
    Renvoie la couleur du gagnant (ROUGE ou JAUNE), ou 0 si pas de gagnant.
    '''
    # 1. Vérification Verticale (dans une colonne)
    # On s'arrête à LIGNE - 3 car on vérifie i, i+1, i+2, i+3
    for c in range(NOMBRE_DE_COLONNE):
        for l in range(NOMBRE_DE_LIGNE - 3):
            if plateau[c][l] == plateau[c][l+1] == plateau[c][l+2] == plateau[c][l+3] and plateau[c][l] != 0:
                return plateau[c][l]
    # 2. Vérification Horizontale (sur une ligne)
    # On s'arrête à COLONNE - 3 car on vérifie c, c+1, c+2, c+3
    for c in range(NOMBRE_DE_COLONNE - 3):
        for l in range(NOMBRE_DE_LIGNE):
            if plateau[c][l] == plateau[c+1][l] == plateau[c+2][l] == plateau[c+3][l] and plateau[c][l] != 0:
                return plateau[c][l]
    # 3. Diagonale Montante ( / )
    # On va vers la droite (c+1) et vers le haut (l+1)
    for c in range(NOMBRE_DE_COLONNE - 3):
        for l in range(NOMBRE_DE_LIGNE - 3):
            if plateau[c][l] == plateau[c+1][l+1] == plateau[c+2][l+2] == plateau[c+3][l+3] and plateau[c][l] != 0:
                return plateau[c][l]
    # 4. Diagonale Descendante ( \ )
    # On va vers la droite (c+1) et vers le bas (l-1)
    # Attention, l'indice de ligne doit commencer à 3 pour pouvoir descendre
    for c in range(NOMBRE_DE_COLONNE - 3):
        for l in range(3, NOMBRE_DE_LIGNE):
            if plateau[c][l] == plateau[c+1][l-1] == plateau[c+2][l-2] == plateau[c+3][l-3] and plateau[c][l] != 0:
                return plateau[c][l]
    return 0 # Personne n'a gagné pour l'instant

def statut_jeu(plateau):
    '''
    array(array(entier)) --> chaine de caracteres
    Renvoie l'état actuel de la partie en texte.
    '''
    victoire = gagnant(plateau)

    if victoire == ROUGE:
        return "Les rouges gagnent"
    elif victoire == JAUNE:
        return "Les jaunes gagnent"
    elif plateau_rempli(plateau):
        return "Partie nulle"
    else:
        return "La partie continue"


# Partie Aléatoire
# Partie ordi versus humain
# Pour chaque position de l'arbre, mettre le compteur (nb de passages par la case, victoire rouge, victoire jaune, égalité)
# Savoir alimenter les compteurs pour chaque nouvelle partie (voir feuille)

import random

def nombre_de_joueur(plateau, k):
    return np.count_nonzero(plateau == k)

def simuler_partie_aleatoire(): #ordi vs ordi
    '''
    Simule une partie de Puissance 4 entre deux joueurs aléatoires (ordi vs ordi).
    Renvoie le plateau final et la durée de la partie.
    '''
    plateau = generer()
    print("Début de la partie aléatoire !")
    afficher_plateau(plateau)

    temps_debut = time.time() # Démarre le chronomètre

    while statut_jeu(plateau) == "La partie continue":
        joueur = joueur_suivant(plateau)
        possibilites = mouvement_possible(joueur, plateau)
        if not possibilites:
            break
        plateau = random.choice(possibilites) # L'ordinateur choisit un plateau au hasard parmi les possibilités
        couleur = "ROUGE" if joueur == ROUGE else "JAUNE"
        print(f"\nLe joueur {couleur} a joué :")
        afficher_plateau(plateau)

    temps_fin = time.time() # Arrête le chronomètre
    duree_partie = temps_fin - temps_debut

    print("\n--- RÉSULTAT FINAL ---")
    print(statut_jeu(plateau))
    print(f"Durée totale de la partie aléatoire : {duree_partie:.4f} secondes.")
    return plateau, duree_partie


#test
'''
print("\n--- Simulation de partie aléatoire (Ordi vs Ordi) ---")
plateau_final_aleatoire, duree_aleatoire = simuler_partie_aleatoire()
print(f"Le statut final de la partie aléatoire est : {statut_jeu(plateau_final_aleatoire)}")
print(f"La durée totale de la partie aléatoire est : {duree_aleatoire:.4f} secondes.")
'''
# fin test


# À faire pour la prochaine fois
# Pour chaque position de l'arbre, mettre le compteur (nb de passages par la case, victoire rouge, victoire jaune, égalité)
# Savoir rétro-alimenter les compteurs pour chaque nouvelle partie (voir feuille)
# Voir wikipedia recherche arborescence Monte Carlo
# Faire une fonction (plateau+compteur) à l'aide d'un dico




def simuler_partie_aleatoire_avec_plateau(plateau):
	'''
	array(array(entier)) --> int, array(array(entier))
	prend un plateau de jeu en entrée, joue une partie aléatoire et renvoie la couleur du gagnant et le plateau final.
	'''
	plateau_courant = np.copy(plateau) # Travaille sur une copie pour éviter de modifier le plateau d'entrée
	while statut_jeu(plateau_courant) == "La partie continue":
		joueur = joueur_suivant(plateau_courant)
		possibilites = mouvement_possible(joueur, plateau_courant)
		if not possibilites:
			break
		plateau_courant = random.choice(possibilites) # L'ordinateur choisit un plateau au hasard parmi les possibilités

	return (gagnant(plateau_courant), plateau_courant)

#test simuler_partie_aleatoire_avec_plateau
#un_plateau = genererbis()
#simuler_partie_aleatoire_avec_plateau(un_plateau)
#fin test




def algorithme_machine(plateau) :
  '''
  array(array(entier)) --> dict(array(array(entier)) : (int,int,int))
  prend un plateau et renvoie un dictionnaire dont les clés sont les
  mouvements possibles et les valeurs sont des triplets, respectivement
  le nombre de victoires rouges, le nombre de victoires jaunes et le nombre de parties nulles.
  '''
  dico_parametre = {}
  # Déterminer quel joueur fait le mouvement depuis 'plateau'
  joueur_faisant_mouvement = joueur_suivant(plateau) # C'est le joueur dont nous évaluons les mouvements

  for mouvements in mouvement_possible(joueur_faisant_mouvement, plateau) :
    nb_partie_gagner = 0
    nb_partie_perdu = 0
    nb_partie_nul = 0
    for i in range(NOMBRE_DE_PARTIE_ALEATOIRE) :
      plateau_memoriser = np.copy(mouvements) # Faire une copie pour ne pas modifier le mouvement original
      gagnant_partie, plateau_partie = simuler_partie_aleatoire_avec_plateau(plateau_memoriser) # Ne déballe plus la valeur du temps
      # Vérifier le résultat par rapport au joueur qui a fait 'mouvements' (joueur_faisant_mouvement)
      if gagnant_partie == joueur_faisant_mouvement :
        nb_partie_gagner += 1
      elif gagnant_partie == 0 :
        nb_partie_nul += 1
      else : # L'adversaire a gagné
        nb_partie_perdu += 1

    # Attribuer nb_victoire_rouge et nb_victoire_jaune en fonction du 'joueur_faisant_mouvement'
    if joueur_faisant_mouvement == ROUGE :
      nb_victoire_rouge = nb_partie_gagner
      nb_victoire_jaune = nb_partie_perdu
    else : # joueur_faisant_mouvement == JAUNE
      nb_victoire_rouge = nb_partie_perdu
      nb_victoire_jaune = nb_partie_gagner
    dico_parametre[tuple(map(tuple, mouvements))] = (nb_victoire_rouge,nb_victoire_jaune,nb_partie_nul)
  return dico_parametre




#test algorithme_machine
'''
un_plateau = genererbis()
resultat_algo = algorithme_machine(un_plateau)
print("Affichage détaillé des résultats de algorithme_machine :\n")
for cle_tuple, v in resultat_algo.items():
    # Convertir la clé (tuple de tuples) en un tableau NumPy pour l'affichage
    plateau_pour_affichage = np.array(cle_tuple)
    print("Mouvement (clé) :")
    afficher_plateau(plateau_pour_affichage)
    print(f"Scores (victoires rouges, victoires jaunes, nulles) : {v}\n---\n")
'''
#fin test


def meilleur_mouvement(plateau) :
  '''
  array(array(entier)) --> array(array(entier))
  prend un plateau de jeu en entrée et renvoie le meilleur mouvement en utilisant
  l'algorithme_machine.
  '''
  dico_parametre_mouvements = algorithme_machine(plateau)

  joueur_actuel = joueur_suivant(plateau) # Le joueur qui doit jouer maintenant

  meilleur_coefficient = -1.0
  meilleur_plateau_choisi = None

  for mouvement_tuple, parametre in dico_parametre_mouvements.items():
    nb_victoire_rouge, nb_victoire_jaune, nb_partie_nulle = parametre

    total_simulations = nb_victoire_rouge + nb_victoire_jaune + nb_partie_nulle

    if total_simulations == 0:
      # Si aucune simulation n'a été faite, traiter ce mouvement comme très mauvais
      coefficient_du_prof = -1.0
    else:
      # Le coefficient est basé sur les victoires pour le joueur actuel
      if joueur_actuel == ROUGE:
        victoires_joueur_actuel = nb_victoire_rouge
      else: # joueur_actuel == JAUNE
        victoires_joueur_actuel = nb_victoire_jaune

      # Calcul du coefficient: victoires / total_simulations
      coefficient_du_prof = victoires_joueur_actuel / total_simulations

    # Mettre à jour si ce mouvement a un meilleur coefficient
    if coefficient_du_prof > meilleur_coefficient:
      meilleur_coefficient = coefficient_du_prof
      meilleur_plateau_choisi = np.array(mouvement_tuple) # Reconvertir en tableau NumPy pour le retour

  return meilleur_plateau_choisi



# Test de la fonction meilleur_mouvement
'''
plateau_initial = genererbis()

print("--- Plateau initial ---")
afficher_plateau(plateau_initial)
print("\n")

meilleur_coup_trouve = meilleur_mouvement(plateau_initial)

if meilleur_coup_trouve is not None:
    print("--- Meilleur mouvement suggéré par l'algorithme ---")
    afficher_plateau(meilleur_coup_trouve)
    # Pour trouver la colonne jouée, on peut comparer le plateau initial et le meilleur coup
    diff = np.where(plateau_initial != meilleur_coup_trouve)
    if diff[0].size > 0:
        colonne_jouee = diff[0][0]
        joueur_place = meilleur_coup_trouve[colonne_jouee, diff[1][0]]
        couleur_joueur = "ROUGE" if joueur_place == ROUGE else "JAUNE"
        print(f"Le meilleur mouvement est de placer un pion {couleur_joueur} dans la colonne : {colonne_jouee}")
else:
    print("Aucun mouvement n'a pu être suggéré (le plateau est peut-être plein ou il n'y a pas de coups valides).")
'''
#fin test


def simuler_partie_machine_vs_machine():
  '''
  none --> none
  Simule une partie de Puissance 4 entre deux "IA".
  Chaque IA utilise l'algorithme 'meilleur_mouvement' pour choisir son coup.
  Affiche l'état du jeu après chaque coup et demande si la partie doit continuer.
  '''
  plateau_jeu = genererbis()
  print("--- Début de la partie Machine vs Machine ---")
  afficher_plateau(plateau_jeu)

  continuer_partie = 'oui'

  while statut_jeu(plateau_jeu) == "La partie continue" and continuer_partie.lower() == 'oui':
    joueur_actuel = joueur_suivant(plateau_jeu)
    couleur_joueur_texte = "ROUGE" if joueur_actuel == ROUGE else "JAUNE"

    print(f"\nC'est au tour du joueur {couleur_joueur_texte} (IA).")

    meilleur_coup = meilleur_mouvement(plateau_jeu)

    if meilleur_coup is None:
      print("Aucun coup valide n'a pu être trouvé. Fin de partie inattendue.")
      break

    plateau_jeu = meilleur_coup

    print(f"Le joueur {couleur_joueur_texte} a joué :")
    afficher_plateau(plateau_jeu)

    statut = statut_jeu(plateau_jeu)
    if statut != "La partie continue":
      print(f"\n--- FIN DE LA PARTIE ---\n{statut}")
      break

    continuer_partie = input("Voulez-vous que la partie continue (oui/non) ? ")

  if continuer_partie.lower() != 'oui':
    print("\nPartie interrompue par l'utilisateur.")
  elif statut_jeu(plateau_jeu) == "La partie continue": # Si la boucle s'est arrêtée pour une autre raison que la victoire/nul ou l'utilisateur
    print("\nFin de partie (raison inconnue ou plateau plein sans gagnant).")

# Appel de la fonction pour lancer une partie
#simuler_partie_machine_vs_machine()


def simuler_partie_humain_vs_machine():
  '''
  none --> none
  Simule une partie de Puissance 4 entre un joueur humain et "IA".
  L'humain choisit sa couleur et joue en entrant la colonne.
  L'IA utilise l'algorithme 'meilleur_mouvement' pour choisir son coup.
  Affiche l'état du jeu après chaque coup et demande si la partie doit continuer.
  '''
  plateau_jeu = genererbis()
  print("--- DÉBUT DE LA PARTIE Humain vs Machine ---")

  # Demander à l'humain de choisir sa couleur
  choix_couleur_valide = False
  while not choix_couleur_valide:
    try:
      humain = int(input("Choisissez votre couleur (1 pour ROUGE, 2 pour JAUNE) : "))
      if humain in [ROUGE, JAUNE]:
        choix_couleur_valide = True
      else:
        print("Veuillez entrer 1 pour ROUGE ou 2 pour JAUNE.")
    except ValueError:
      print("Saisie invalide. Veuillez entrer un nombre.")

  ia = JAUNE if humain == ROUGE else ROUGE

  print(f"Vous jouez les {'ROUGE' if humain == ROUGE else 'JAUNE'}. L'IA joue les {'ROUGE' if ia == ROUGE else 'JAUNE'}.")

  afficher_plateau(plateau_jeu)

  continuer_partie = 'oui'

  while statut_jeu(plateau_jeu) == "La partie continue" and continuer_partie.lower() == 'oui':
    joueur_actuel = joueur_suivant(plateau_jeu)
    couleur_joueur_texte = "ROUGE" if joueur_actuel == ROUGE else "JAUNE"

    if joueur_actuel == humain:
      print(f"\nC'est à votre tour, joueur {couleur_joueur_texte} !")
      colonne_choisie = -1

      # Boucle pour s'assurer que la colonne est valide et non pleine
      while True:
        try:
          colonne_input = input(f"Dans quelle colonne jouer (0 à {NOMBRE_DE_COLONNE-1}) ? (Ou 'q' pour quitter) ")
          if colonne_input.lower() == 'q':
            continuer_partie = 'non'
            break

          colonne_choisie = int(colonne_input)

          if not (0 <= colonne_choisie < NOMBRE_DE_COLONNE):
            print(f"Saisie invalide. Entrez un nombre entre 0 et {NOMBRE_DE_COLONNE-1}.")
          elif colonne_pleine(plateau_jeu[colonne_choisie]):
            print("Cette colonne est pleine ! Choisissez-en une autre.")
          else:
            # Trouver la première case vide dans la colonne choisie
            for ligne in range(NOMBRE_DE_LIGNE):
              if plateau_jeu[colonne_choisie][ligne] == 0:
                plateau_jeu[colonne_choisie][ligne] = humain
                break
            break # Sortir de la boucle de choix de colonne

        except ValueError:
          print("Saisie invalide. Veuillez entrer un nombre ou 'q'.")

      if continuer_partie.lower() == 'non': # L'utilisateur a choisi de quitter
        break

    else: # Tour de l'IA
      print(f"\nC'est au tour du joueur {couleur_joueur_texte} (IA).")
      coup_ia = meilleur_mouvement(plateau_jeu)

      if coup_ia is None:
        print("L'IA n'a pas pu trouver de coup valide. Fin de partie inattendue.")
        break

      plateau_jeu = coup_ia
      print(f"Le joueur {couleur_joueur_texte} (IA) a joué :")

    afficher_plateau(plateau_jeu)

    statut = statut_jeu(plateau_jeu)
    if statut != "La partie continue":
      print(f"\n--- FIN DE LA PARTIE ---\n{statut}")
      break

    # Demander à l'utilisateur s'il veut continuer si la partie n'est pas déjà finie
    if statut == "La partie continue":
        continuer_partie = input("Voulez-vous que la partie continue (oui/non) ? ")

  if continuer_partie.lower() != 'oui':
    print("\nPartie interrompue par l'utilisateur.")
  elif statut_jeu(plateau_jeu) == "La partie continue":
    print("\nFin de partie (raison inconnue ou plateau plein sans gagnant).")

# Appel de la fonction pour lancer une partie
#simuler_partie_humain_vs_machine()

def algo_machine_vs_aleatoire():
    '''
    None --> str, float
    Simule une partie de Puissance 4 où un joueur est l'algorithme (meilleur_mouvement)
    et l'autre joueur est aléatoire. Chronomètre la durée de la partie.
    '''
    plateau_jeu = genererbis()
    print("--- Début de la partie Algorithme vs Aléatoire ---")
    afficher_plateau(plateau_jeu)

    temps_debut_partie = time.time()

    # Pour cet exemple, supposons que ROUGE est l'algorithme et JAUNE est aléatoire.
    joueur_algo = ROUGE
    joueur_aleatoire = JAUNE

    print(f"L'algorithme joue en tant que {'ROUGE' if joueur_algo == ROUGE else 'JAUNE'}.")
    print(f"Le joueur aléatoire joue en tant que {'ROUGE' if joueur_aleatoire == ROUGE else 'JAUNE'}.")


    while statut_jeu(plateau_jeu) == "La partie continue":
        joueur_actuel = joueur_suivant(plateau_jeu)
        couleur_joueur_texte = "ROUGE" if joueur_actuel == ROUGE else "JAUNE"

        if joueur_actuel == joueur_algo:
            print(f"\nC'est au tour de l'algorithme ({couleur_joueur_texte}).")
            meilleur_coup = meilleur_mouvement(plateau_jeu)

            if meilleur_coup is None:
                print("L'algorithme n'a pas pu trouver de coup valide. Fin de partie inattendue.")
                break
            plateau_jeu = meilleur_coup
            print(f"L'algorithme ({couleur_joueur_texte}) a joué :")
        else: # joueur_actuel == joueur_aleatoire
            print(f"\nC'est au tour du joueur aléatoire ({couleur_joueur_texte}).")
            possibilites = mouvement_possible(joueur_aleatoire, plateau_jeu)
            if not possibilites:
                print("Le joueur aléatoire n'a pas de coups valides. Fin de partie inattendue.")
                break
            plateau_jeu = random.choice(possibilites)
            print(f"Le joueur aléatoire ({couleur_joueur_texte}) a joué :")

        afficher_plateau(plateau_jeu)

    temps_fin_partie = time.time()
    duree_partie = temps_fin_partie - temps_debut_partie

    statut_final = statut_jeu(plateau_jeu)
    print(f"\n--- FIN DE LA PARTIE ---")
    print(statut_final)
    print(f"Durée totale de la partie : {duree_partie:.4f} secondes.")

    return statut_final, duree_partie

#test
# Pour lancer une partie :
statut, duree = algo_machine_vs_aleatoire()
print(f"Le statut final de la partie est : {statut}")
print(f"La durée totale de la partie est : {duree:.4f} secondes.")
```
